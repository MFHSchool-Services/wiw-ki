Der Seed in Zeile 3 sorgt dafür, dass derselbe Zufall herauskommt. Deine Zahlen können trotzdem in der letzten Stelle etwas von denen im Buch abweichen.

**Zuerst:** Diese Zelle lädt die Fotos von einem schnellen Server. Der Originalserver ist manchmal sehr langsam (bis zu einer halben Stunde). Das dauert hier nur wenige Sekunden.

```python
import keras
# Holt CIFAR-10 vom schnellen Google-Server statt vom
# langsamen Originalserver. Danach findet load_data()
# die Fotos schon fertig vor.
keras.utils.get_file("cifar-10-batches-py-target", extract=True,
    origin="https://storage.googleapis.com/tensorflow/"
           "tf-keras-datasets/cifar-10-python.tar.gz")
print("Fotos sind da.")
```

```python
import keras
from keras import layers
keras.utils.set_random_seed(42)
(bilder, nummern), (testbilder, testnummern) = \
    keras.datasets.cifar10.load_data()
bilder, testbilder = bilder / 255, testbilder / 255
netz = keras.Sequential([
    keras.Input(shape=(32, 32, 3)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")])
netz.compile(optimizer="adam", metrics=["accuracy"],
             loss="sparse_categorical_crossentropy")
print(netz.count_params(), "Stellschrauben")
netz.fit(bilder, nummern, epochs=5)
netz.evaluate(testbilder, testnummern)
```

## Dein eigenes Foto

Führe zuerst die Zelle oben aus, damit das Netz trainiert ist. Dann wählst du hier ein Foto aus. Das Netz kennt nur seine zehn Klassen und muss immer eine davon wählen.

Lade nur Fotos von Dingen, Tieren oder Fahrzeugen hoch, keine Fotos von Menschen.

```python
import numpy as np
from google.colab import files
klassen = ["Flugzeug", "Auto", "Vogel", "Katze", "Hirsch",
           "Hund", "Frosch", "Pferd", "Schiff", "Lastwagen"]
name = list(files.upload())[0]
foto = keras.utils.load_img(name, target_size=(32, 32))
werte = netz.predict(np.array([foto]) / 255)
print(klassen[werte.argmax()])
```
