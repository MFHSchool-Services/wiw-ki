Du lädst 60 000 kleine Fotos (CIFAR-10), baust ein Faltungsnetz und trainierst es in fünf Durchgängen. Beim ersten Mal lädt Colab die Fotos herunter, das dauert einen Moment.

Klicke links neben der grauen Code-Zelle auf das Dreieck ▶ oder drücke Strg + Enter (am Mac Cmd + Enter).

Der Seed in Zeile 2 sorgt dafür, dass derselbe Zufall herauskommt. Deine Zahlen können trotzdem in der letzten Stelle etwas von denen im Buch abweichen.

```python
import keras
keras.utils.set_random_seed(42)
keras.utils.get_file("cifar-10-batches-py-target",
    "https://storage.googleapis.com/tensorflow/tf-keras-"
    "datasets/cifar-10-python.tar.gz", extract=True)
(bilder, nummern), (testbilder, testnummern) = \
    keras.datasets.cifar10.load_data()
netz = keras.Sequential([
    keras.Input(shape=(32, 32, 3)),
    keras.layers.Conv2D(32, (3, 3), activation="relu"),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Conv2D(64, (3, 3), activation="relu"),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Flatten(),
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dense(10, activation="softmax")])
netz.compile(optimizer="adam", metrics=["accuracy"],
             loss="sparse_categorical_crossentropy")
print(netz.count_params(), "Stellschrauben")
netz.fit(bilder / 255, nummern, epochs=5)
netz.evaluate(testbilder / 255, testnummern)
```

## Dein eigenes Foto

Führe zuerst die Zelle oben aus, damit das Netz trainiert ist. Dann wählst du hier ein Foto aus. Das Netz kennt nur seine zehn Klassen und muss immer eine davon wählen.

Lade nur Fotos von Dingen, Tieren oder Fahrzeugen hoch, keine Fotos von Menschen.

```python
import numpy as np
from google.colab import files
klassen = ["Flugzeug", "Auto", "Vogel", "Katze", "Hirsch",
           "Hund", "Frosch", "Pferd", "Schiff", "Lastwagen"]
name = list(files.upload())[0]
foto = keras.utils.load_img(name, target_size=(32, 32))
werte = netz.predict(np.array([foto]) / 255)
print(klassen[werte.argmax()])
```
