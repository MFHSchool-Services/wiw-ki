# Eine vortrainierte KI nutzen

**Tipp:** Du kannst alles ändern und ausprobieren. Wenn du deine Änderungen behalten willst, wähle oben *Datei → Kopie in Drive speichern*.

Du lädst MobileNetV2, eine vortrainierte KI, die schon an über einer Million Fotos gelernt hat, 1 000 Dinge zu unterscheiden. Dann lässt du sie ein eigenes Foto erkennen.

Klicke links neben der grauen Code-Zelle auf das Dreieck ▶ oder drücke Strg + Enter (am Mac Cmd + Enter).

Lade nur Fotos von Dingen, Tieren oder Fahrzeugen hoch, keine Fotos von Menschen.

```python
import keras, numpy as np
from keras.applications import mobilenet_v2 as mn
from google.colab import files

ki = mn.MobileNetV2(weights="imagenet")
print(ki.count_params(), "Stellschrauben")

name = list(files.upload())[0]
foto = keras.utils.load_img(name, target_size=(224, 224))
x = keras.utils.img_to_array(foto)
x = mn.preprocess_input(np.array([x]))
tipps = mn.decode_predictions(ki.predict(x), top=3)[0]
for _, wort, p in tipps:
    print(wort, round(p * 100, 1), "%")
```

**Probier’s aus:** Führe die Zelle mit weiteren Fotos aus. Fotografiere dasselbe Ding einmal von nah und einmal von weit weg.
