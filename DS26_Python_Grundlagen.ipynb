Dein erstes Programm: Text ausgeben, mit Variablen rechnen, Listen, eine Funktion und eine Schleife benutzen und aus den Zahlen einer KI ihren Tipp machen.

Klicke links neben der grauen Code-Zelle auf das Dreieck ▶ oder drücke Strg + Enter (am Mac Cmd + Enter).

```python
print("Hallo KI!")                       # zeigt Text an
punkte = 500                             # eine Variable
print("Punkte:", punkte + 1)

klassen = ["Katze", "Hund", "Frosch"]    # eine Liste
werte = [0.62, 0.3, 0.08]                # Antwort der KI
print(klassen[0], klassen[2])            # Index ab 0

def prozent(zahl):                       # eine Funktion
    return round(zahl * 100, 1)

for i in range(3):                       # eine Schleife
    print(klassen[i], prozent(werte[i]), "%")

beste = werte.index(max(werte))
print("Tipp der KI:", klassen[beste])
```
